```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("qoute_dataset.csv")
df
```

```python
quotes = df["quote"]
quotes.head()
```

```python
# lets convert words into lowercase

quotes = quotes.str.lower()
quotes.head()
```

```python
# lets remove , .  and all from text
import string

translator = str.maketrans('','',string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
quotes.head()
```

```python
# doing tokenization now:

from tensorflow.keras.preprocessing.text import Tokenizer

vocab_size = 10000
tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(quotes)
sequences = tokenizer.texts_to_sequences(quotes)
```

```python
# means we have 8979 diff words that means 10000 is good amount
word_index = tokenizer.word_index
print(len(word_index))
```

```python
# for sentence completion we want to create smt like if we write 714 rord then next suggesstion would be 63
# for 63 30 for 30 20 and if we write 2,3etc words together lets say 713 63 30 then next word sugg would be 20
# so this is what out output will look like and we will have many combinations

for i in range(3):
    print(sequences[i])
```

```python
X = []
y = []

for seq in sequences:
    for i in range(1,len(seq)):
        input_seq = seq[:i]
        output_seq = seq[i]
        X.append(input_seq)
        y.append(output_seq)

# now we have got out inputs and outputs but lets bring them on length
```
